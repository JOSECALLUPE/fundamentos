1. Escribe un programa que pida al usuario la base y la altura de un triángulo y
calcule su área.

```python
print("--- Cálculo del Área de un Triángulo ---")

base = float(input("Ingresa la base del triángulo: "))


altura = float(input("Ingresa la altura del triángulo: "))


area = (base * altura) / 2

print(f"El área del triángulo es: {area}")
```

2. Escribe un programa que lea una temperatura en grados Celsius e imprima su
equivalente en Fahrenheit.


```python
print("\n--- Conversor de Temperaturas (Celsius a Fahrenheit) ---")

celsius = float(input("Ingresa la temperatura en grados Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print(f"{celsius}°C equivalen a {fahrenheit}°F")
```

3. Escribe un programa que reciba tres números diferentes e indique cuál es el mayor.

```python
print("\n--- Encontrar el Mayor de Tres Números ---")

num1 = float(input("Ingresa el primer número: "))
num2 = float(input("Ingresa el segundo número: "))
num3 = float(input("Ingresa el tercer número: "))

if num1 >= num2 and num1 >= num3:
    mayor = num1
elif num2 >= num1 and num2 >= num3:
    mayor = num2
else:
    mayor = num3

print(f"El mayor de los tres números es: {mayor}")
```

4. Un cliente realiza una compra y se le aplica un descuento según el monto total:
• Si gasta más de S/500, se le aplica 20%.
• Si gasta entre S/200 y S/500, se le aplica 10%.
• Si gasta menos de S/200, no tiene descuento

```python
print("\n--- Cálculo de Descuento por Compra ---")

monto_total = float(input("Ingresa el monto total de la compra (en Soles): S/"))

descuento = 0
monto_final = 0

if monto_total > 500:
    descuento = monto_total * 0.20
    monto_final = monto_total - descuento
    print("¡Felicidades! Se te aplica un 20% de descuento.")
elif monto_total >= 200 and monto_total <= 500:
    descuento = monto_total * 0.10
    monto_final = monto_total - descuento
    print("Se te aplica un 10% de descuento.")
else:
    monto_final = monto_total
    print("Lo siento, tu compra no aplica para descuento.")

print(f"Monto original: S/{monto_total:.2f}")
print(f"Descuento aplicado: S/{descuento:.2f}")
print(f"Monto final a pagar: S/{monto_final:.2f}")
```

5. Solicita dos números enteros al usuario (inicio y fin) y muestra la suma de todos
los números pares en ese rango (inclusive).

```python
print("\n--- Suma de Números Pares en un Rango ---")

inicio = int(input("Ingresa el número de inicio del rango (entero): "))
fin = int(input("Ingresa el número de fin del rango (entero): "))

suma_pares = 0

if inicio > fin:
    inicio, fin = fin, inicio

print(f"Calculando la suma de números pares entre {inicio} y {fin} (inclusive):")

for numero in range(inicio, fin + 1):

    if numero % 2 == 0:
        suma_pares += numero

print(f"La suma de los números pares en el rango es: {suma_pares}")
```
